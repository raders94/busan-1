# Series

- Series는 인덱스가 붙어있는 1차원 배열 데이터
- 엑셀에서 열과 동일


```python
import pandas as pd

scores = pd.Series([45, 86, 23, 67, 12])


print(type(scores))
```

```python
scores = pd.Series([45, 86, 23, 67, 12])

len(scores)
print(scores.ndim)  #차원
print(scores.shape) #모양
print(scores.size)  #크기
print(scores.values) #값
print(scores.index) #인덱스
```

```python
print(scores[0]) #첫번째 값
```

```python
scores = pd.Series([45, 86, 23, 67, 12], index=['A', 'B', 'C', 'D', 'E'])

print(scores)
print(scores['A'])
```

```python
sales = pd.Series([100, 120, 150], index=['서울', '서울', '부산'])

print(sales.ndim)  #차원
print(sales.shape) #모양
print(sales.size)  #크기
print(sales.values) #값
print(sales.index) #인덱스

print(sales['서울']) #서울에 대한 값
```

```python
sales.index.is_unique #중복 값이 있는지 확인
```

```python
s= pd.Series([3.2, 5.1, 9.0], index=['1월', '2월', '3월'], name='평균 기온')

s
```

```python
mixed= pd.Series([False,  0])
mixed
```

```python
dtype_series = pd.Series([1, 2, 3, 4, 5], dtype='float64')
dtype_series
```

```python
float_series = pd.Series([64.0, 10.2, 3.0],dtype='float64')
float_series.astype('int64')

```

```python
list_numbers = [10, 20, 30]

s = pd.Series(list_numbers)
s*2
s > 15
```

```python
s[s > 15] #masking, 필터
```

```python
a= pd.Series([1, 2, 3], index=['x', 'y', 'z'])
b= pd.Series([10, 20, 30], index=['z', 'y', 'x'])
```

```python
a+b
```

```python
a= pd.Series([1, 2], index=['x', 'y'])
b= pd.Series([10, 20], index=['z', 'y'])

a+b  #NaN은 실수
```

```python
data = {'서울': 100, '부산': 200, '대구': 300}
s = pd.Series(data, index=['서울', '부산', '광주'])
s
```

```python
#Series의 속성

scores = pd.Series([45, 86, 23, 67, 12], index=['A', 'B', 'C', 'D', 'E'], name='성적' dtype='int64')
scores.values  # 값들
scores.index  # 인덱스
scores.name  # Series 이름
```
