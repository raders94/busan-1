```python
arrival = pd.to_datetime(df['입항일시'], errors='coerce')
arrival
```

```python
departure = pd.to_datetime(df['출항일시'], errors='coerce')
departure
```

```python
stay = (departure - arrival).dt.total_seconds()/3600
stay
```

```python
stay.describe()
```

```python
valid = arrival.dt.year.eq(df['입항일시']) & stay.between(0,24*30)

stay[valid]
```

```python
date_cols = ['입항일시', '출항일시', '출항예정일시']

for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors='coerce')

df['체류시간'] = (df['출항일시'] - df["입항일시"]).dt.total_seconds()/3600
df['출항지연시간'] = (df['출항일시'] - df["출항예정일시"]).dt.total_seconds()/3600

df[ ['체류시간', '출항지연시간']]

valid_stay = (df['입항일시'].dt.year.eq(df['입항연도'])
              & df['체류시간'].between(0,24*30))

valid_schedule = valid_stay & (df['출항예정일시'].between(df['입항일시'] - pd.Timedelta(days=1),
                                                    df['입항일시'] - pd.Timedelta(days=30))) #timedelta 오차범위

valid_stay.sum()
            
```

```python
pd.Series({
    '원본 행 수': len(df),
    '입항연도와 입항일시 연도 불일치': df['입항일시'].dt.year.ne(df['입항연도']).sum(),
    '음수 체류시간' : df['체류시간'].lt(0).sum(),
    '30일 초과': df['체류시간'].gt(24*30).sum(),
    '유효 체류시간 비율': valid_stay.mean()*100,
    '완전중복행' : df.duplicated().sum()
})
```

```python
stay_quantile = df.loc[valid_stay, '체류시간'].quantile([0.5,0.75,0.9,0.95,0.99])

stay_quantile.index = ['중앙값', '75백분위', '90백분위', '95백분위', '99백분위']

stay_quantile
```

```python
print(f'24시간 이내 출항 비율= {df.loc[valid_stay, '체류시간'].le(24).mean():.2f}')
print(f'7일 초과 체류 비율= {df.loc[valid_stay, '체류시간'].gt(24*7).mean():.2f}')
```
