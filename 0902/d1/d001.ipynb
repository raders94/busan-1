```python
terminal_col =[]

for col_name in df.columns:
    if col_name not in ['년도', '합계']:
        terminal_col.append(col_name)

print(terminal_col)
```

```python
new_port=[]

for col_name in df.columns:
    if col_name.startswith('신항'):
        new_port.append(col_name)

print(new_port)
```

```python
df_1 = df.copy()

df[new_port].sum(axis=1)
#df[new_port].T.sum()
```

```python
df_1['부두합계'] = df_1[terminal_col].T.sum()
df_1['신항합계'] = df_1[new_port].T.sum()
df_1['신항비중'] = df_1['신항합계']/df_1['부두합계']

df_1[['년도', '부두합계', '신항합계', '신항비중']]
```
