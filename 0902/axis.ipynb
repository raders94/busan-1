```python
import pandas as pd

orders = pd.DataFrame(
    {
        "order_date": pd.to_datetime([
            "2026-08-01",
            "2026-08-02",
            "2026-08-03",
            "2026-08-04",
            "2026-08-05",
            "2026-08-06",
        ]),
        "customer": [
            "민준", "서연", "지우",
            "도윤", "하준", "수빈",
        ],
        "region": [
            "서울", "부산", "서울",
            "대전", "부산", "서울",
        ],
        "product": [
            "키보드", "모니터", "마우스",
            "키보드", "노트북", "모니터",
        ],
        "quantity": [2, 1, 3, 2, 1, 2],
        "unit_price": [
            45000,
            320000,
            25000,
            45000,
            1350000,
            320000,
        ],
        "paid": [
            True, True, False,
            True, True, False,
        ],
    },
    index=pd.Index(
        ["O001", "O002", "O003", "O004", "O005", "O006"],
        name="order_id",
    ),
)

orders["total_amount"] = (
    orders["quantity"] * orders["unit_price"]
)

orders
```

```python
orders.index
```

```python
orders.columns
```

```python
orders.loc['O002'] #행만 가저올수 있음 #1차원 자료구조
```

```python
orders.loc['O002', 'customer'] #행과 열을 동시에 가져올수 있음
```

```python
orders.loc[['O001', 'O003'], ['customer', 'product']
] # 2차원 자료구조
```

```python
orders.loc['O002' : 'O005']
```

```python
orders.loc['O002' : 'O005', 'region': 'unit_price']
```

```python
orders.iloc[1:3, 0:3] #iloc 위치기반, 끝 포함
```

```python
orders[1:3] #iloc 위치기반, 끝 미포함
```

```python
orders.at['O004', 'product'] #at은 단일값 접근에 최적화 되어있음
```

```python
orders[(orders['quantity'] >2)& (orders['total_amount'] > 100000)][['product', 'quantity']]
```
