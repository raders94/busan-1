## if - else - elif

else if 와 같이 if절 평가 내용이 거짓일 경우 다른 조건을 추가하고자 할때 사용


```python
score = 85
grade = "F"

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C" 
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"점수: {score}, 학점: {grade}")
```

## 반복문

반복문은 특정 코드 블록을 조건이 만족되거나 특정 횟수 만큼 실행하는 제어구조

## #while

주어진 조건식이 참인 동안에 코드 블럭을 반복해서 실행하는 제어구조

- while 문 구조
'''python
while 조건식 or 값:
    반복형_명령문
'''


```python
is_python_easy = True

while is_python_easy:
    print("파이썬은 쉬워요!")
    break  # 무한루프 방지

count = 0
while count < 5:
    print(f"카운트: {count}")
    count += 1
print("루프 종료")
```

### for문

'''python
for 꺼내온 요소를 담을 변수 in 시퀀스 자료형:
    명령문
'''

```python
for number in range(5):
    print(f"숫자: {number}")
```

```python
fruits = ["사과", "바나나", "딸기"]
for fruit in fruits:
    print(f"과일: {fruit}")
```

### break

현재 실행중인 루프를 즉시 탈출

```python
for i in range(10):
    if i==5:
        print(f'{i}')
        break
    print(f'현재 숫자: {i}')
```
