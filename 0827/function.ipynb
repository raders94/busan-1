```python
i=0
j=6
while i <=6:
    if i<=4 and i!=2:
        print(" " * (4-i) + "*" * (2*i+1))
        i +=1
        j -=1
    else:
        print(" " * (4-j) + "*" * (2*j+1))
        i +=1
        j -=1

```
