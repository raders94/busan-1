```python
month = [1,2,3,4,5,6]
product_a = [ 128, 135, 155, 133, 555, 234]
product_d = [ 90, 235, 55, 233, 355, 434]

plt.plot(month, product_a)
plt.plot(month, product_d)

plt.show()
```

```python
month = [1,2,3,4,5,6]
product_a = [ 128, 135, 155, 133, 555, 234]
product_d = [ 90, 235, 55, 233, 355, 434]

plt.plot(month, product_a, label = 'product_a')
plt.plot(month, product_d, label = 'product_d')

plt.legend()
plt.show()
```

```python
month = [1,2,3,4,5,6]
product_a = [ 128, 135, 155, 133, 555, 234]
product_d = [ 90, 235, 55, 233, 355, 434]
product_c = [ 34, 65, 75, 12, 75, 64]

plt.plot(month, product_a, label = 'product_a')
plt.plot(month, product_d, label = 'product_d')
plt.plot(month, product_c, label = 'product_c')

plt.legend()
plt.show()
```

```python
month = [1,2,3,4,5,6]
product_a = [ 128, 135, 155, 133, 555, 234]

title = 'monthly sales'

plt.plot(month, product_a)
plt.title(title,fontsize=18, 
          #loc='left'
          )
plt.xlabel('month')
plt.ylabel('sales')

print(plt.gca().get_title())
print(plt.gca().get_xlabel())

plt.show()
```

```python
month = [1,2,3,4,5,6]
product_a = [ 128, 135, 155, 133, 186, 234]

plt.plot(month,product_a, marker='o') # 
plt.xlim(0, 7)
plt.ylim(90,200)  
plt.xticks([1,2,3,4],['JANUARY','FEBRuARY','MARCH','APRIL'],rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
A=[10,15,20,25,30]
B=[12,17,22,27,32]
C=[9,14,19,24,29]

plt.plot(x,A, color='green', label = 'A') #RED, BLUE, GREEN, ORANGE, BLACK
plt.plot(x,B, color='black', label = 'B')
plt.plot(x,C, color='blue', label = 'C')

plt.legend()

plt.show()
```
