# 연산자

연산자는 특정한 계산이나 처리를 수행하도록 지시하는 기호 또는 키워드를 의미

연산자에 의해서 처리되는 대상이 되는 값아너 변수를 피연산자라고 한다.

## 연산자의 구분
연산자는 자신이 처리하는 피연사의 갯수에 따라서 구분이 가능하다.

### 단항연산자
단항연산자는 하나의 피연산자에 대해서 연산을 수행하는 연산자들을 의미

### 이항연산자
이항연산자 두개의 피연산자에 대해서 연산을 수행하는 연사자들을 의미
