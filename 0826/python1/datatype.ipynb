```python
fruits = ["사과", "바나나", "포도", "딸기"]

print(fruits[0]) #단건조회

print(fruits[1:3]) #범위조회

print(fruits[0:3:2])
print(fruits[::-1])

print(fruits[0][1])

fruits.append("망고")
print(fruits)
```
