# Series의 이해

Series에 저장된 여러 값을 범위로 선택하고, 조건에 맞는 값만 추출하며, 값을 수정하거나 연산하는 방법

```python
import pandas as pd
import numpy as np

sales = pd.Series([100, 200, 300, 400, 500], 
                  index=['월', '화', '수', '목', '금'], 
                  name='매출액', dtype='int64')

sales
```

```python
sales.iloc[1:4]
```

```python
sales.iloc[3:100]  # 범위를 벗어난 인덱스는 무시됨
```

```python
sales.iloc[10:20]
```

```python
sales.loc['월':'수']  # 라벨 기반 슬라이싱
```

```python
sales.loc['월']
sales.loc['화']
sales.loc['수']

sales[['월', '수', '금']]
```

```python
sales[:'목']  # 라벨 기반 슬라이싱
```

# Series 조건 필터링

조건에 맞는 데이터들만 필터링해서 가져오는 방법

```python
scores = pd.Series(
    [88, 95, 62, 70, 100,  95,  58], 
    index=['학생1', '학생2', '학생3', '학생4', '학생5', '학생6', '학생7'],
    name='성적', dtype='int64'
)
```

## 조건 필터링

Pandas의 Series는 비교연산자 연산 지원

```python
scores >=80  #boolean Series 반환
```

```python
scores[(scores >= 80) & (scores <= 90)]  # 비트 연산자 & : and, | : or, ~ : not
```

```python
scores[(scores >= 80) | (scores <= 90)]
```

## between, isin



```python
scores.between(80, 90)  # between() 메서드로 범위 조건을 만족하는 값 선택
scores.isin([62, 70, 100])  # isin() 메서드로 특정 값 선택
```

```python
scores[scores.between(80, 90, inclusive=True)]  # inclusive=True : 양 끝값 포함, False : 양 끝값 제외
scores[scores.between(80, 90, inclusive='right')]  # inclusive=right : 오른쪽 끝값 포함, 왼쪽 끝값 제외
scores[scores.between(80, 90, inclusive='left')]  # inclusive=left : 왼쪽 끝값 포함, 오른쪽 끝값 제외
```

```python
scores = pd.Series([45, 86, 23, 67, 12], index=['A', 'B', 'C', 'D', 'E'], name='성적', dtype='int64')

print(f'합격자 수 = {(scores >= 60).sum()}명') # 합격자 수
print(f'합격률 = {(scores < 60).mean()}명') # mean() 메서드로 평균 계산, True=1, False=0으로 계산되어 합격률 계산 가능
print(f'한명이라도 만점이 있는가 = {(scores == 100).any()}') # any() 메서드로 하나라도 True가 있는지 확인
print(f'모두 합격했는가 = {(scores >= 60).all()}') # all() 메서드로 모두 True인지 확인
```
