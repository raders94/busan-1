### 기술통계의 목적

- 중심 경향치 : 데이터가 어디에 위치 => 평균, 중앙값 확인
- 최빈값 : 가장 높은 빈도
- 산포도 : 데이터의 분산 확인
 - 분산 : (평균-수치)^2의 평균
 - 표준편차 = std = 분산^1/2
 - 사분위수 : 전체 데이터를 4등분 위치의 값

- 분포의 형태
 - 왜도 : 분포의 치우친 정도
 - 첨도 : 분포의 뾰족한 정도
