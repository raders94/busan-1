```python
import numpy as np

products = ['A', 'B', 'C']
sales_2026=[120,160,140]
sales_2025=[140,168,170]


x= np.arange(len(products))

width = 0.35

print(f'center={x-width/2}')
print(f'left:{x+width/2}')


fig, ax=plt.subplots(figsize=(8,4))

ax.bar(
    x - width/2,
    sales_2025,
    width,
    label='2025'
)
ax.bar(
    x + width/2,
    sales_2026,
    width,
    label='2026'
)

ax.set_xticks(x)
ax.set_xticklabels(products)

ax.set_title('Sales by year')
ax.set_xlabel('products')
ax.set_ylabel('sales')


ax.legend()
plt.show()
```

```python
import numpy as np

products = ['A', 'B', 'C']
sales_2026=[120,160,140]
sales_2025=[140,168,170]
sales_2024=[170,138,140]


x= np.arange(len(products))

width = 0.20

print(f'center={x}')
print(f'left:{x+width}')
print(f'right:{x+width}')

fig, ax=plt.subplots(figsize=(10,4))

ax.bar(
    x - width,
    sales_2024,
    width,
    label='2024'
)
ax.bar(
    x,
    sales_2025,
    width,
    label='2025'
)

ax.bar(
    x + width,
    sales_2026,
    width,
    label='2026'
)




ax.set_xticks(x)
ax.set_xticklabels(products)

ax.set_title('Sales by year')
ax.set_xlabel('products')
ax.set_ylabel('sales')


ax.legend()
plt.show()
```

```python
df=pd.DataFrame({
    'study_hours': [2,4,6,2.5,5.5,4,5,7,8],
    'scores': [60,67,65,45,86,86,59,56,45,],
    'class' : ['A','A','A','A','A','B','B','B','B']
})
fig, ax =plt.subplots(figsize=(8,6))

class_a=df[df['class']=='A']
class_b=df[df['class']=='B']

ax.scatter(
    class_a['study_hours'],
    class_a['scores'],
    label='ClassA'
)

ax.scatter(
    class_b['study_hours'],
    class_b['scores'],
    label='ClassB'
)

ax.set_xlabel('study hours')
ax.set_ylabel('scores')

ax.legend()
plt.show()
```

```python
import numpy as np

rng = np.random.default_rng(42)
x= rng.normal(
    loc=50,
    scale=10,
    size=500
)
y= (x*0.7+rng.normal(loc=0,scale=8,size=500))
fig, ax = plt.subplots(figsize=(8,6))

ax.scatter(x,y, s=40, alpha=0.45) #alpha 반투명화=> 겹친거 확인 가능

plt.show()
```
